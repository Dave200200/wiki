# Test Berechnung

## Formel

$$Q = U \cdot A \cdot \Delta T$$

Wobei:
- $Q$ = Wärmestrom [W]
- $U$ = Wärmedurchgangskoeffizient [W/m²K]
- $A$ = Fläche [m²]
- $\Delta T$ = Temperaturdifferenz [K]

```python
# Eingabewerte
U = 0.3      # W/m²K
A = 10.0     # m²
delta_T = 20 # K

# Berechnung
Q = U * A * delta_T

print(f"Wärmestrom Q = {Q} W")
```

## Hinweise

- Norm: EN ISO 6946
- U-Wert je nach Bauteil und Dämmung verschieden
- Ergebnis gilt für stationären Wärmedurchgang
