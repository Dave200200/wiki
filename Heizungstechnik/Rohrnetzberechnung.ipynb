## Berechnung Massen- und Volumenströme

### Volumenstrom aus der Fluidleistung

$$\dot{Q}_{fl} = \dot{m} \cdot c_p \cdot (\vartheta_{VL} - \vartheta_{RL}) = \dot{V} \cdot \rho \cdot c_p \cdot (\vartheta_{VL} - \vartheta_{RL})$$

$$\boxed{\dot{V} = \frac{\dot{Q}_{fl}}{\rho \cdot c_p \cdot (\vartheta_{VL} - \vartheta_{RL})}}$$

### Strömungsgeschwindigkeit

$$\dot{V} = A \cdot v \xrightarrow{\text{kreisrund}} \frac{d_i^2 \cdot \pi}{4} \cdot v$$

$$\boxed{v = \frac{4 \cdot \dot{V}}{d_i^2 \cdot \pi}}$$

## Ermittlung der Druckverluste

### 1. Gerades Rohr

$$\Delta p_R = \lambda \cdot \frac{l}{d_i} \cdot \frac{\rho}{2} \cdot v^2 \quad [Pa]$$

In der Praxis auch häufig:

$$\Delta p_R = R \cdot \ell$$

mit dem längenbezogenen Druckabfall $R$ in Pa/m (aus Herstellertabellen).

Bei Vorprojektplanung mit Zuschlag für Einzelwiderstände:

$$\Delta p_{R+E} = 1{,}3 \text{ bis } 2{,}0 \cdot R \cdot \ell$$

(Faktor 1,3 bei einfachen Anlagen mit langen Strecken, Faktor 2,0 bei vielen Umlenkungen)

### 2. Formstücke (Bögen, T-Stücke, Winkel)

$$\Delta p_E = \sum \zeta \cdot \frac{\rho}{2} \cdot v^2 \quad [Pa]$$

Die $\zeta$-Werte werden aus Tabellenbüchern oder Herstellerangaben entnommen.

### 3. Geräte und Armaturen

$$\Delta p_{GuA} = \left(\frac{\dot{V}}{k_{VS}}\right)^2 \cdot 10^5 \quad \text{[Pa]}$$

oder alternativ mit Bezugsvolumenstrom:

$$\Delta p_{GuA} = \left(\frac{\dot{V}}{\dot{V}_{bez}}\right)^2 \cdot \Delta p_{bez}$$

## Ermittlung der Rohrreibungszahl

Die Reynolds-Zahl bestimmt die Strömungsform:

$$Re = \frac{v \cdot d_i}{\nu}$$

Für Wasser gilt näherungsweise (nach Glück, 1985):

$$\nu_{H_2O} = 19{,}8 \cdot 10^{-6} \cdot \vartheta^{-0{,}915} \quad \text{(gültig für } 40 < \vartheta < 200\text{°C)}$$

### Laminare Strömung
($Re \leq 2320$)

$$\lambda = \frac{64}{Re}$$

### Turbulente Strömung – hydraulisch glatt

Bedingung: $\frac{v \cdot \epsilon}{\nu} < 65$

**Blasius** ($2320 < Re \leq 10^5$):

$$\lambda = \frac{0{,}3164}{\sqrt[4]{Re}}$$

**Nikuradse** ($Re > 10^5$):

$$\lambda = 0{,}0032 + \frac{0{,}221}{Re^{0{,}237}}$$

**Prandtl-Karman** ($Re > 2320$, implizit):

$$\frac{1}{\sqrt{\lambda}} = 2 \cdot \log(Re \cdot \sqrt{\lambda}) - 0{,}8$$

### Turbulente Strömung – hydraulisch rau 
Bedingung:$\frac{v \cdot \epsilon}{\nu} \geq 65$

**Prandtl-Colebrook-White** (universelles Widerstandsgesetz):

$$\frac{1}{\sqrt{\lambda}} = -2 \cdot \log\left(\frac{2{,}51}{Re \cdot \sqrt{\lambda}} + \frac{\epsilon}{3{,}71 \cdot d_{i,h}}\right)$$

Als Startwert für die Iteration (explizite Näherung):

$$\lambda = \frac{0{,}25}{\left[\log\left(\frac{\epsilon}{3{,}71 \cdot d_{i,h}} + \frac{5{,}74}{Re^{0{,}9}}\right)\right]^2}$$
