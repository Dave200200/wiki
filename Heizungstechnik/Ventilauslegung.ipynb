# Ventilauslegung

## Allgemeine Kennwerte

Neben den hydraulischen Auslegungsgrößen sind bei der Ventilauswahl folgende allgemeine Kenndaten zu beachten:

- **Medium:** Wasser, Dampf, Sole etc.
- **Nenndruck:** PN 6, 10, 16, 25 …
- **Maximal zulässige Mediumtemperatur**

## Hydraulische Stellglieder
Die Auslegung hydraulischer Stellglieder (Regelventile) ist ein zentraler Schritt in der Planung von Heizungs- und Kälteanlagen. Ein korrekt ausgelegtes Ventil gewährleistet, dass die Anlage im gesamten Lastbereich regelbar bleibt und der Wärmeüberträger die gewünschte Leistung abgibt.
### Anforderungen an das Stellglied

Ein Stellglied muss im hydraulischen Kreis folgende Bedingungen erfüllen:

- **Linearität:** Möglichst linearer Zusammenhang zwischen Ventilhub und der am Wärmeabgabesystem abgegebenen Wärmeleistung
- **Regelbereichsausnutzung:** Ausnutzung des gesamten verfügbaren Regelbereichs
- **Mindestleistung:** Möglichst kleine, minimal regelbare Wärmeleistung

---

### Der kv-Wert

Der **kv-Wert** (bzw. **kvs-Wert** für das voll geöffnete Ventil) ist die wichtigste Kenngröße zur Auslegung eines Regelventils. Er ist definiert als der Volumenstrom in m³/h, der bei einer Druckdifferenz von **1 bar (= 100 kPa)** über dem voll geöffneten Ventil durchfließt.

#### Berechnung

Aus dem kv-Wert lässt sich der Druckabfall über das Ventil berechnen:

$$\Delta p_{GuA} = \left(\frac{\dot{V}}{k_{VS}}\right)^2 \cdot 10^5 \quad [Pa]$$

Umgekehrt ergibt sich der Volumenstrom bei bekanntem Druckabfall und kvs-Wert:

$$\dot{V} = k_{VS} \cdot \sqrt{\frac{\Delta p_{GuA}}{10^5}}$$

### Ventilautorität

Die **Ventilautorität** $a_V$ beschreibt, welchen Anteil der gesamten Druckdifferenz im Regelkreis das Ventil übernimmt:

$$a_V = \frac{\Delta p_{V100}}{\Delta p_{V100} + \Delta p_{MV}}$$

| Größe | Bedeutung |
|---|---|
| $\Delta p_{V100}$ | Druckdifferenz am voll geöffneten Ventil beim Nennvolumenstrom |
| $\Delta p_{MV}$ | Druckdifferenz im mengenvariablen Rohrnetz (ohne Ventil) |

> **Praxisempfehlung:** $a_V \geq 0{,}5$
>
> Liegt die Ventilautorität darunter, verliert das Ventil zunehmend seinen Einfluss auf den Volumenstrom, was die Regelgüte deutlich verschlechtert. Die Druckdifferenz über dem Regelventil sollte mindestens so groß gewählt werden wie die Druckdifferenz über dem mengenvariablen Rohrnetzanteil.


### Grundkennlinien von Stellarmaturen

Die **Grundkennlinie** beschreibt den Zusammenhang zwischen Ventilhub $H$ und dem kv-Wert bei konstantem Differenzdruck (ideale Bedingungen, ohne Anlageneinfluss).

#### Lineare Kennlinie

Bei gleicher Änderung des Ventilhubs ergibt sich die gleiche **absolute** Änderung des kv-Werts:

$$k_v = k_{v100} \cdot \frac{H}{H_{100}}$$

#### Gleichprozentige Kennlinie

Bei gleicher Hubänderung ergibt sich die gleiche **prozentuale** Änderung des kv-Werts. Die Empfindlichkeit des Ventils steigt mit dem Hub:

$$k_v = \frac{k_{v100}}{e^{n \cdot \left(1 - \frac{H}{H_{100}}\right)}} \quad \text{mit} \quad n = \ln(S_{V0}) = \ln\left(\frac{k_{v100}}{k_{v0}}\right)$$

#### Vergleich der Kennlinien

| Kennlinie | Charakteristik | Typische Anwendung |
|---|---|---|
| Linear | Gleichmäßige kv-Änderung über den Hub | Luftkühler, Vorlauftemperaturregelung |
| Gleichprozentig | Überproportionale kv-Änderung im oberen Hubbereich | Lufterhitzer |


### Stellverhältnis

Das **theoretische Stellverhältnis** $S_{V0}$ beschreibt den Regelbereich des Ventils unter Idealbedingungen:

$$S_{V0} = \frac{k_{V100}}{k_{V0}}$$

In der Praxis ist das **praktische Stellverhältnis** $S_{Vr}$ relevant, da die reale Anlagenkennlinie vom Idealfall abweicht.

### Volumenstromkennlinie unter Berücksichtigung der Ventilautorität

Unter Berücksichtigung der Ventilautorität $a_V$ ergibt sich der tatsächlich durch den Heizkreis fließende Volumenstrom $\dot{V}(H)$:

$$\frac{\dot{V}(H)}{\dot{V}_{100}} = \frac{1}{\sqrt{1 + a_V \cdot \left(\left(\frac{kv_{100}}{kv(H)}\right)^2 - 1\right)}}$$

Je kleiner die Ventilautorität, desto stärker weicht die resultierende Anlagenkennlinie von der idealen Grundkennlinie ab.


### Maximale Druckdifferenz $\Delta p_{max}$

Die maximale Druckdifferenz $\Delta p_{max}$ bezeichnet die maximal zulässige Druckdifferenz am **geschlossenen** Ventil. Wird dieser Wert überschritten, kann der Stellantrieb das Ventil nicht mehr dicht geschlossen halten.

> Dieser Wert ist zwingend aus dem Herstellerdatenblatt zu entnehmen und mit den Betriebsbedingungen der Anlage abzugleichen.

### Leckverluste

Leckverluste $\dot{V}_L$ werden vom Hersteller in Prozent des kvs-Werts angegeben:

| Pfad | Typische Leckrate |
|---|---|
| Durchgangspfad | 0,02 – 0,05 % von kvs |
| Bypasspfad | bis zu 2 % von kvs |

Bei **Vierwegemischern** treten durch die vollständige Trennung von Erzeuger- und Verbraucherkreis keine Leckverluste auf.


### Ventilauswahl

#### Kennlinienauswahl nach Anwendungsfall

| Anwendungsfall | Empfohlene Kennlinie |
|---|---|
| Heizungsanlage mit Vorlauftemperaturregelung | Linear oder gleichprozentig/linear |
| Regelstrecke mit Luftkühler | Linear |
| Regelstrecke mit Lufterhitzer | Gleichprozentig oder lineares Tellerventil |

> Die **Regelstreckenkennlinie** – also der Zusammenhang zwischen Wärmeleistung und Ventilhub – ergibt sich aus der Kombination von Volumenstromkennlinie und Wärmeübertragungsverhalten ($a_W$-Wert). Ziel ist eine möglichst konstante Verstärkung über den gesamten Regelbereich.

#### Wahl der Nennweite

Die Nennweite der Stellarmatur ist so zu wählen, dass die listenmäßige Druckdifferenz am Ventil gleich oder geringfügig größer als die berechnete Druckdifferenz ist.

> **Faustformel:** Die Ventilnennweite liegt häufig **eine Dimension unter** der Rohrnennweite.

#### Standardisierte kvs-Werte (Renard'sche Reihe)

Ventile sind in standardisierten kvs-Stufen erhältlich:

| kvs [m³/h] | 1,6 | 2,5 | 4 | 6,3 | 10 | 16 | 25 | 40 | 63 | 100 | 160 |
|---|---|---|---|---|---|---|---|---|---|---|---|
| DN [mm] | 15 | 15 | 15(20) | 20 | 25 | 32 | 40 | 50 | 65 | 80 | 100 |

Herstellerabhängig sind für eine Nennweite mehrere kvs-Werte verfügbar, um die Auslegung an den berechneten Nennvolumenstrom und die vorhandene Druckdifferenz anpassen zu können

### Zusammenfassung: Auslegungsablauf

1. Nennvolumenstrom $\dot{V}_{100}$ aus Heizlast und Temperaturspreizung berechnen
2. Verfügbare Druckdifferenz am Ventileinbauort bestimmen
3. Erforderlichen kvs-Wert berechnen: $k_{VS} = \frac{\dot{V}}{\sqrt{\Delta p / 10^5}}$
4. Passende Nennweite und kvs-Stufe aus dem Herstellerkatalog auswählen
5. Ventilautorität prüfen: $a_V \geq 0{,}5$
6. Maximale Druckdifferenz am geschlossenen Ventil gegen $\Delta p_{max}$ des Herstellers prüfen
7. Geeignete Grundkennlinie für die Regelstrecke festlegen
