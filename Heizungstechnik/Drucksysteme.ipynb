# Drucksysteme

## Druckhaltung geschlossene Systeme

### 1. Statischer Druck
statische Druck: höchster Punkt der Anlage - Anschlusspunkt des Druckhaltesystems (empfohlen: Saugseite der Umwälzpumpe)
$$p_{st} = \frac{h_{st}}{10}$$

$h_{st}$ = statische Höhe [m]  
$p_{st}$ = statischer Druck [bar]

---

### 2. Dampfdruck $p_v$
Der Dampfdruck (als Überdruck) bei maximaler Betriebstemperatur der Anlage. Muss berücksichtigt werden, damit an keiner Stelle der Anlage das Wasser siedet.
$$p_v = p_{v,\text{abs}} - p_{atm} \qquad (p_{atm} \approx 1{,}013\ \text{bar})$$

| Temperatur | $p_v$ (Überdruck) |
|---|---|
| 70 °C  | ≈ −0,7 bar (praktisch 0) |
| 90 °C  | ≈ −0,3 bar (praktisch 0) |
| 100 °C | 0 bar |
| 110 °C | ≈ 0,4 bar |
| 120 °C | ≈ 1,0 bar |
| 130 °C | ≈ 1,69 bar |

**Hinweis:** Negative Werte werden in der Berechnung als $p_v = 0$ eingesetzt (vernachlässigt)

---

### 3. Mindestbetriebsdruck $p_0$
Mindestbetriebsdruck muss mindestens Summe aus statischem Druck, Dampfdruck und 0,2 bar Sicherheitszuschlag entsprechen (ÖNorm EN 12827)
=> Kavitationsschutz

$$p_0 = p_{st} + p_v + 0{,}2\ \text{bar}$$

Bei Druckhaltesystemen, die **nicht** auf der Saugseite der Pumpe sitzen, muss zusätzlich der Druckanstieg der Umwälzpumpe berücksichtigt werden.

---

### 4. Anfangsdruck $p_{ini}$ (Fülldruck)
Niedrigster Druck im Betriebsbereich der Anlage (Kaltzustand, Pumpe aus). An jeder Stelle der Anlage muss der Überdruck $\geq 0.5$ bar sein.
$$p_{ini} = p_0 + 0{,}3\ \text{bar}$$
$$p_{ini} = p_{stat} + 0{,}5\ \text{bar}$$

Der Vordruck des Membran-Ausdehnungsgefäßes wird auf denselben Wert eingestellt: $p_{\text{Vordruck, MAG}} = p_{im}$.

**Kontrolle am höchsten Punkt:**

$$p_{ini} - p_{st} \geq 0{,}5\ \text{bar}$$

---

### 5. Enddruck $p_e$
Höchster zulässiger Druck im Betriebsbereich (MAG maximal komprimiert), begrenzt durch den Ansprechdruck des Sicherheitsventils $p_{SV}$.
$$p_e = p_{SV} - 0{,}5\ \text{bar}$$

---

### 6. Arbeitsbereich des Ausdehnungsgefäßes

$$\Delta p = p_e - p_{ini}$$

Je kleiner $\Delta p$, desto größer muss das MAG-Volumen sein, um dieselbe Wasserausdehnung aufzunehmen.

---

### Quelle

ÖNORM EN 12828, Anhang D – *Hinweise für die Auslegung von Membran-Druckausdehnungsgefäßen und Druckhaltesystemen (geschlossene Systeme)*, Abschnitt D.1 (a–f).

## Dimensionierung Membran-Ausdehnungsgefäß (MAG)

### 1. Ausdehnungsvolumen $V_e$

Das Ausdehnungsvolumen ist die Wassermenge, die durch thermische Ausdehnung zusätzlich entsteht.

$$V_e = V_{Anlage} \times n$$

- $V_{Anlage}$ = Wasserinhalt der gesamten Anlage [Liter]
- $n$ = Ausdehnungskoeffizient des Wassers [-] (abhängig von der max. Betriebstemperatur, bezogen auf ca. 10 °C Kaltwasser)

**Ausdehnungskoeffizient $n$ (Richtwerte):**

| Max. Temperatur | $n$ |
|---|---|
| 30 °C | 0,0043 |
| 34 °C | ≈ 0,0050 |
| 40 °C | 0,0079 |
| 50 °C | 0,0121 |
| 60 °C | 0,0171 |
| 70 °C | 0,0227 |
| 80 °C | 0,0290 |
| 90 °C | 0,0359 |

---

### 2. Nennvolumen des Ausdehnungsgefäßes $V_n$

$$V_n = V_e \times \frac{p_e + 1}{p_e - p_0}$$

- $p_e$ = Enddruck [bar] (Überdruck)
- $p_0$ = Mindestbetriebsdruck [bar] (Überdruck)
- Der Term $(p_e + 1)$ approximiert den Bezug auf den Absolutdruck ($p_{atm} \approx 1$ bar)

**Interpretation:** Je kleiner die Differenz $(p_e - p_0)$, desto größer wird $V_n$ – ein enger Druckkorridor erfordert also ein größeres Gefäß, um dieselbe Ausdehnung aufzunehmen.

---

### 3. Praktische Auslegungsgröße (Nutzinhalt vs. Nennvolumen)

Das am Markt erhältliche Gefäß muss auf den nächstgrößeren **Normwert** aufgerundet werden (z. B. 25 / 35 / 50 / 80 / 100 / 150 / 200 Liter), da Hersteller nur diskrete Baugrößen anbieten.

$$V_n^{gewählt} \geq V_n^{berechnet}$$

---

### 4. Berechnungsschema – Übersicht

| Schritt | Formel | Ergebnis |
|---|---|---|
| 1. Ausdehnungskoeffizient $n$ bestimmen | Tabelle, nach max. Temperatur | – |
| 2. Ausdehnungsvolumen | $V_e = V_{Anlage} \times n$ | Liter |
| 3. Nennvolumen | $V_n = V_e \times \dfrac{p_e + 1}{p_e - p_0}$ | Liter |
| 4. Gefäß wählen | nächstgrößerer Normwert | Liter |

---

### 5. Benötigte Eingangsgrößen

- **$V_{Anlage}$** – Wasserinhalt der Anlage [L] (Rohrleitungen, Wärmetauscher, Kühler, Verteiler etc.)
- **Max. Betriebstemperatur** – zur Bestimmung von $n$
- **$p_0$, $p_e$** – aus der Druckhaltungs-Berechnung (siehe Kapitel *Druckhaltung geschlossener Systeme*)

---

### 6. Beispiel (Fortsetzung Kühlsystem, 14 m statische Höhe)

Angenommen $V_{Anlage} = 2000\ \text{L}$, max. Temp. 34 °C:

$$V_e = 2000 \times 0{,}005 = 10\ \text{L}$$

Mit $p_0 = 1{,}6$ bar, $p_e = 2{,}5$ bar:

$$V_n = 10 \times \frac{2{,}5 + 1}{2{,}5 - 1{,}6} = 10 \times \frac{3{,}5}{0{,}9} \approx 38{,}9\ \text{L}$$

→ **gewähltes Gefäß: 50 Liter** (nächster Normwert)

---

### Quelle

ÖNORM EN 12828 / ÖNORM H 5195-1 – Auslegung von Membran-Druckausdehnungsgefäßen in geschlossenen Wasserheizungs-/Kühlanlagen.
