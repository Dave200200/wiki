
## Hydraulischer Abgleich

Die einwandfreie Funktion einer Heizanlage hängt davon ab, dass jede Raumheizeinrichtung im Betrieb den in der Planung errechneten Heizwasserstrom erhält. Dies erfordert nach der Montage und Befüllung eine sorgfältige Einregulierung – den hydraulischen Abgleich.

> **Merke:** Die Aufgabe des hydraulischen Abgleichs besteht darin, die richtigen Volumenströme für die jeweiligen Anlagenteile (Heizkörper, Kessel, Luftheizregister usw.) unter Nennbedingungen zu gewährleisten. Das Fluid geht den Weg des geringeren Widerstandes – bei Anlagenteilen mit geringeren Widerständen fließt mehr Wasser hindurch als bei solchen mit höheren Widerständen.

### Abgleichmöglichkeiten

- Wahl anderer (kleinerer) Rohrweiten
- Einsatz eines Heizkörperventils mit einem anderen (kleineren) $k_v$-Wert
- Einsatz eines Heizkörperventils mit fester Voreinstellmöglichkeit
- Einsatz einer fest einstellbaren Rücklaufverschraubung
- Einsatz von Strangregulierventilen
- Einsatz von Differenzdruckregelventilen

### Strangregulierventile

**An jedem Kessel und an jeder Kühlmaschine:**
- Einregulierung der richtigen Wassermengenverteilung zwischen Kesseln/Kühlmaschinen
- Sicherstellung, dass kein Kessel/keine Kühlmaschine zu große oder zu kleine Durchflussmengen erhält
- Überprüfung, dass auf Wärmeerzeugungsseite und Verteilerseite gleich große Energiemengen transportiert werden

**Bei allen Verteilern, Steigsträngen und Abzweigen:**
- Einregulierung der richtigen Durchflussmengen
- Überprüfung der Vergleichbarkeit der transportierten Energiemengen zwischen Verteilungskreisen und einzelnen Zonen

**In allen Zonen mit einer Sekundärpumpe:**
- Abgleich von überdimensionierten Regelventilen, Sekundärpumpen und Wärmeverbrauchern
- Überprüfung der Gleichheit der transportierten Energiemengen zwischen Verteilungssystem und einzelnen Kreisen

**Strangregulierventile in Verbindung mit Dreiwegverteilventilen:**
- Eines in Serie mit dem Regelventil zur Einstellung der Nenndurchflussmengen
- Eines im Bypass zur Sicherstellung gleicher Durchflussmengen bei voll geöffnetem oder vollständig geschlossenem Ventil

**Bypass-Druckausgleichsventil** für jeden Strang mit Zweiwegeregelventilen (z.B. Thermostatventile):
- Zur Stabilisierung des Differenzdruckes über das Regelventil

> **Heizkörper oder Einzelraumkühleinheiten** müssen mit voreinstellbaren Ventilen oder Rücklaufverschraubungen ausgestattet werden, um die für jeden Wärmeverbraucher individuell benötigte Durchflussmenge einregulieren zu können.

### Dimensionierung der Einregulierungsventile

Bei der Dimensionierung von Strangregulierventilen gilt:

$$\Delta p_{SRV} \geq 3000\, Pa$$

Die Einbaulängen vor und nach dem Ventil sind zu beachten:
- Ohne Pumpe: 2D vor dem Ventil, 5D nach dem Ventil
- Mit Pumpe: 2D vor dem Ventil, 10D nach dem Ventil

*(D = Ventil-Nenndurchmesser DN)*
